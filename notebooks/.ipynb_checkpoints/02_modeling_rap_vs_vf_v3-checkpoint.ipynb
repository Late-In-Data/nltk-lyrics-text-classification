# 02 — De l’observation à la prédiction : classifier des paroles (Rap vs Variété Française)

Dans le notebook d’EDA, nous avons observé des différences nettes entre les paroles de Rap et de Variété Française (longueur, vocabulaire, motifs récurrents, effets “artiste”).  
Ici, on passe du constat à la question suivante : **peut-on prédire le genre d’une chanson uniquement à partir de ses paroles ?**

Ce notebook met en place un pipeline de **classification de textes** (prétraitement → TF-IDF → modèles scikit-learn), puis compare plusieurs approches, avant d’interpréter les résultats (mots discriminants, analyse d’erreurs, limites).

---

## Plan
1. Chargement du dataset consolidé (`songs_master.csv`)
2. Baselines
3. Modèles “simples mais solides” (NB / LogReg / LinearSVC)
4. Optimisation rapide (RandomizedSearchCV)
5. Interprétation (mots discriminants + erreurs typiques)
6. (Option) Évaluation plus robuste “par artiste”


```python
# Imports
import re
import numpy as np
import pandas as pd
from pathlib import Path

from sklearn.model_selection import train_test_split, RandomizedSearchCV, GroupShuffleSplit
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay, accuracy_score, f1_score

from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC

import matplotlib.pyplot as plt

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
```

## 1) Chargement des données

On part d’un fichier **consolidé** (créé dans l’EDA) afin d’éviter de relire chaque CSV individuellement.

Chemin attendu : `data/processed/songs_master.csv`

```python
from pathlib import Path

# Détection d'un "project root" (utile si le notebook est dans /notebooks)
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()

# On privilégie le CSV consolidé (créé en EDA), sinon fallback
candidates = [
    PROJECT_ROOT / 'data' / 'processed' / 'songs_master.csv',
    PROJECT_ROOT / 'data' / 'processed' / 'data_song.csv',
    PROJECT_ROOT / 'data' / 'data_song.csv',
]
for p in candidates:
    if p.exists():
        path = p
        break
else:
    raise FileNotFoundError('Impossible de trouver le CSV master. Cherché: ' + ', '.join(str(c) for c in candidates))

data_song = pd.read_csv(path)
print('📄 Fichier chargé :', path)
print('Shape :', data_song.shape)
print('\nRépartition des classes :\n', data_song['genre'].value_counts())
data_song.head()
```

### 1.1 Nettoyage minimal

On conserve `lyrics` (texte brut). Le nettoyage fin sera fait via le `preprocessor` dans le vectorizer (lowercase, suppression d’URLs, normalisation espaces).

```python
data_song = data_song.dropna(subset=["lyrics", "genre"]).copy()
data_song["lyrics"] = data_song["lyrics"].astype(str)
data_song = data_song[data_song["genre"].isin(["Rap", "VF"])].copy()

data_song["genre"].value_counts()
```

## 2) Stopwords (NLTK + fichier custom)

On utilise les stopwords **français NLTK** + ton fichier `data/stopword.txt`.

> Conseil : garde exactement cette cellule (ou très proche) dans EDA et modélisation pour rester cohérent.

```python
import nltk
from nltk.corpus import stopwords as nltk_stopwords

try:
    _ = nltk_stopwords.words("french")
except LookupError:
    nltk.download("stopwords")

STOP = set(nltk_stopwords.words("french"))

custom_sw_path = Path(r"C:\Users\latej\Desktop\Projets_github\nltk-lyrics-text-classification\data\stopword.txt")
if custom_sw_path.exists():
    STOP |= set(custom_sw_path.read_text(encoding="utf-8").split())

print("Nb stopwords :", len(STOP))
```

## 3) Split train/test

Split stratifié classique. L’option “split par artiste” est proposée plus bas pour tester la généralisation.

```python
X = data_song["lyrics"]
y = data_song["genre"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=RANDOM_STATE, stratify=y
)

print("Train:", len(X_train), "Test:", len(X_test))
print("Train distrib:", y_train.value_counts(normalize=True).round(3).to_dict())
print("Test  distrib:", y_test.value_counts(normalize=True).round(3).to_dict())
```

## 4) Vectorisation TF-IDF + modèles

On utilise un `Pipeline` scikit-learn (TF-IDF → modèle). C’est la meilleure pratique (pas de fuite, tuning plus simple).

```python
def preprocessor(text: str) -> str:
    """Nettoyage léger, commun à tous les modèles.
    On évite de faire un nettoyage agressif ici : l'objectif est de rester proche du texte.
    """
    if text is None:
        return ""
    text = str(text).lower()
    text = re.sub(r"http\S+|www\.\S+", " ", text)  # urls
    text = re.sub(r"\s+", " ", text).strip()
    return text

# TF-IDF sur mots (baseline "forte" pour la classification de texte)
tfidf_word = TfidfVectorizer(
    preprocessor=preprocessor,
    stop_words=STOP,
    ngram_range=(1, 2),
    min_df=2
)

# TF-IDF sur caractères (modèle plus robuste aux variations d'orthographe / contractions / argot)
# Remarque : pas de stopwords en char-ngrams.
tfidf_char = TfidfVectorizer(
    preprocessor=preprocessor,
    analyzer='char_wb',
    ngram_range=(3, 5),
    min_df=2
)
```

### 4.1 Baseline : classe majoritaire

Point de comparaison indispensable : prédire la classe majoritaire.

```python
from collections import Counter

major = Counter(y_train).most_common(1)[0][0]
y_pred_major = np.array([major] * len(y_test))

print("Baseline (majoritaire):")
print("  Accuracy :", round(accuracy_score(y_test, y_pred_major), 4))
print("  F1 macro :", round(f1_score(y_test, y_pred_major, average="macro"), 4))
print("  F1 Rap   :", round(f1_score(y_test, y_pred_major, pos_label="Rap"), 4))
```

### 4.2 Comparaison de modèles (mots vs caractères)

On compare des modèles linéaires classiques (très performants sur TF-IDF) ainsi qu'une variante **plus avancée** :
- **TF-IDF mots (1–2 grams)** : interprétable, standard.
- **TF-IDF caractères (3–5 grams)** : souvent très efficace en français (accents, apostrophes, argot, variantes).


```python
models = {
    # TF-IDF mots
    'NB (TF-IDF mots)': ('word', MultinomialNB()),
    'LogReg (TF-IDF mots)': ('word', LogisticRegression(max_iter=2000)),
    'LinearSVC (TF-IDF mots)': ('word', LinearSVC()),

    # Modèle "plus poussé" : TF-IDF caractères + SVM linéaire
    'LinearSVC (TF-IDF caractères)': ('char', LinearSVC())
}

rows = []
fitted = {}

for name, (vect_type, model) in models.items():
    vect = tfidf_word if vect_type == 'word' else tfidf_char
    clf = Pipeline([('tfidf', vect), ('model', model)])
    clf.fit(X_train, y_train)
    fitted[name] = clf

    y_pred = clf.predict(X_test)
    rows.append({
        'Model': name,
        'Vectorisation': 'TF-IDF mots (1-2)' if vect_type == 'word' else 'TF-IDF caractères (3-5)',
        'Accuracy': accuracy_score(y_test, y_pred),
        'F1_macro': f1_score(y_test, y_pred, average='macro'),
        'F1_Rap': f1_score(y_test, y_pred, pos_label='Rap')
    })

res_df = pd.DataFrame(rows).sort_values('F1_macro', ascending=False)
res_df
```

### 4.3 Détails sur le meilleur modèle (sans tuning)

On affiche le **classification report** et la **matrice de confusion** afin de comprendre le type d’erreurs.
Ensuite, on interprète le modèle via les tokens les plus discriminants (quand c’est possible).


```python
best_name = res_df.iloc[0]['Model']
best_clf = fitted[best_name]

y_pred = best_clf.predict(X_test)
print('Meilleur modèle (sans tuning) :', best_name)
print(classification_report(y_test, y_pred, digits=3))

cm = confusion_matrix(y_test, y_pred, labels=['Rap', 'VF'])
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Rap', 'VF'])
disp.plot()
plt.title('Matrice de confusion — meilleur modèle (sans tuning)')
plt.show()
```

## 5) Optimisation rapide (RandomizedSearchCV)

On optimise **uniquement** le meilleur modèle "mots" (standard) pour rester léger et comparable.
Le modèle caractères est gardé comme variante avancée (souvent déjà très bon sans tuning).


```python
# On tune un modèle basé sur TF-IDF mots (plus classique / interprétable)
candidates_to_tune = [m for m in res_df['Model'].tolist() if '(TF-IDF mots)' in m]
to_tune = candidates_to_tune[0] if candidates_to_tune else 'LinearSVC (TF-IDF mots)'

if 'LinearSVC' in to_tune:
    pipe = Pipeline([('tfidf', tfidf_word), ('model', LinearSVC())])
    param_distributions = {
        'tfidf__ngram_range': [(1, 1), (1, 2)],
        'tfidf__min_df': [1, 2, 3, 5],
        'model__C': np.logspace(-2, 1, 20),
    }
else:
    pipe = Pipeline([('tfidf', tfidf_word), ('model', LogisticRegression(max_iter=3000))])
    param_distributions = {
        'tfidf__ngram_range': [(1, 1), (1, 2)],
        'tfidf__min_df': [1, 2, 3, 5],
        'model__C': np.logspace(-2, 1, 20),
        'model__solver': ['liblinear', 'lbfgs'],
    }

search = RandomizedSearchCV(
    pipe,
    param_distributions=param_distributions,
    n_iter=25,
    scoring='f1_macro',
    cv=5,
    random_state=RANDOM_STATE,
    n_jobs=-1,
)

search.fit(X_train, y_train)

print('Modèle tuné :', to_tune)
print('Meilleurs paramètres :', search.best_params_)
print('Score CV (f1_macro) :', round(search.best_score_, 4))

tuned_clf = search.best_estimator_
y_pred_tuned = tuned_clf.predict(X_test)

print('\nPerformance sur le test :')
print('  Accuracy :', round(accuracy_score(y_test, y_pred_tuned), 4))
print('  F1 macro :', round(f1_score(y_test, y_pred_tuned, average='macro'), 4))
print('  F1 Rap   :', round(f1_score(y_test, y_pred_tuned, pos_label='Rap'), 4))

print('\nClassification report (tuned) :')
print(classification_report(y_test, y_pred_tuned, digits=3))

cm = confusion_matrix(y_test, y_pred_tuned, labels=['Rap', 'VF'])
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Rap', 'VF'])
disp.plot()
plt.title('Matrice de confusion — modèle tuné')
plt.show()
```

## 6) Interprétation

Cette section vise à répondre à : **qu’est-ce que le modèle a réellement appris ?**

On regarde :
- les **n-grams** les plus associés à chaque classe (quand le modèle est linéaire et basé sur TF-IDF mots),
- quelques **exemples d’erreurs** pour comprendre les cas limites.

> Remarque : l’interprétation par "mots importants" est surtout pertinente pour la vectorisation **mots**.
Pour la vectorisation **caractères**, l’interprétation est possible mais moins intuitive (n-grams de caractères).


```python
def show_top_tokens(pipeline, top_n=20):
    vec = pipeline.named_steps["tfidf"]
    model = pipeline.named_steps["model"]
    feature_names = np.array(vec.get_feature_names_out())

    if not hasattr(model, "coef_"):
        print("⚠️ Ce modèle n'a pas de coefficients (coef_).")
        return

    classes = model.classes_
    coefs = model.coef_

    # Cas binaire (certaines implémentations -> shape (1, n_features))
    if coefs.shape[0] == 1 and len(classes) == 2:
        coefs = np.vstack([-coefs[0], coefs[0]])

    for i, cls in enumerate(classes):
        top_pos = np.argsort(coefs[i])[-top_n:][::-1]
        print(f"\nClasse: {cls}")
        print("  +:", ", ".join(feature_names[top_pos]))

# Si le modèle tuné n'est pas linéaire, on force une LogReg pour l'interprétation
interp = tuned_clf
if not hasattr(interp.named_steps["model"], "coef_"):
    interp = Pipeline([("tfidf", tfidf), ("model", LogisticRegression(max_iter=3000))]).fit(X_train, y_train)

show_top_tokens(interp, top_n=20)
```

### 6.2 Analyse d’erreurs (exemples)

On affiche quelques **faux positifs** et **faux négatifs** pour comprendre pourquoi le modèle hésite.
Ces exemples sont très utiles pour identifier :
- des paroles VF au style "parlé" (proches du rap),
- des textes Rap plus narratifs/romantiques (proches de la VF),
- l’impact des refrains répétitifs ou de certaines signatures d’écriture.


```python
errors = pd.DataFrame({
    "lyrics": X_test.values,
    "y_true": y_test.values,
    "y_pred": y_pred_tuned
})

fp = errors[(errors["y_true"] == "VF") & (errors["y_pred"] == "Rap")].head(5)
fn = errors[(errors["y_true"] == "Rap") & (errors["y_pred"] == "VF")].head(5)

print("Faux positifs (VF prédit Rap) :")
fp[["y_true", "y_pred", "lyrics"]]
```

```python
print("Faux négatifs (Rap prédit VF) :")
fn[["y_true", "y_pred", "lyrics"]]
```

## 7) (Option) Évaluation plus robuste “par artiste”

Pour éviter d’évaluer un modèle qui apprend uniquement une **signature d’artiste**, on peut faire un split où les artistes du test ne sont pas présents dans le train.
Si la performance baisse, c’est un signal qu’une partie de la prédiction venait de patterns spécifiques à certains artistes — mais si elle reste bonne, cela indique un signal de **genre** plus généralisable.


```python
if 'artist' in data_song.columns:
    groups = data_song['artist']

    gss = GroupShuffleSplit(n_splits=1, test_size=0.30, random_state=RANDOM_STATE)
    train_idx, test_idx = next(gss.split(X, y, groups=groups))

    X_train_g, X_test_g = X.iloc[train_idx], X.iloc[test_idx]
    y_train_g, y_test_g = y.iloc[train_idx], y.iloc[test_idx]

    tuned_clf.fit(X_train_g, y_train_g)
    y_pred_g = tuned_clf.predict(X_test_g)

    print('Split par artiste :')
    print('  Accuracy :', round(accuracy_score(y_test_g, y_pred_g), 4))
    print('  F1 macro :', round(f1_score(y_test_g, y_pred_g, average='macro'), 4))
    print('  F1 Rap   :', round(f1_score(y_test_g, y_pred_g, pos_label='Rap'), 4))
else:
    print("Colonne 'artist' absente du master CSV : option non applicable.")
```

## 8) Conclusion

- Les modèles linéaires sur **TF-IDF** offrent déjà d’excellentes performances sur cette tâche.
- L’optimisation (RandomizedSearchCV) améliore généralement la performance, notamment via le choix des **n-grams** et du paramètre **C**.
- La comparaison **mots vs caractères** permet de tester une approche plus robuste aux variantes d’écriture (argot, apostrophes, orthographe).
- Enfin, l’évaluation **par artiste** est une vérification importante : elle mesure mieux la capacité du modèle à généraliser au-delà des signatures d’artistes.
