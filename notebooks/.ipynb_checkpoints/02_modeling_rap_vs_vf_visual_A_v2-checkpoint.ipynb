# 02 — Modélisation : des paroles à la prédiction (Rap vs Variété Française)

Dans l’EDA, on a observé des différences nettes de style et de vocabulaire entre le Rap et la Variété Française.  
Ici, on passe à la **prédiction** : transformer les paroles en variables numériques (TF‑IDF) puis entraîner des modèles de classification.

Ce notebook met l’accent sur :
- la **comparaison visuelle** des modèles,
- l’**évaluation** (métriques + matrice de confusion),
- l’**interprétation** (tokens discriminants, analyse d’erreurs),
- et une validation plus réaliste via un **split par artiste** (si disponible).


```python
from pathlib import Path
import pandas as pd
import numpy as np
```

## 1) Chargement des données

```python
csv_path = r"C:\Users\latej\Desktop\Projets_github\nltk-lyrics-text-classification\data\data_song.csv"
df = pd.read_csv(csv_path)
print("Shape:", df.shape)
```

```python
print('\nRépartition des classes :\n', df['genre'].value_counts())
```

## 2) Stopwords (NLTK + fichier custom)

On réutilise les mêmes stopwords que dans l’EDA pour rester cohérent.

```python
from nltk.corpus import stopwords as nltk_stopwords
STOP = set(nltk_stopwords.words("french"))
custom_path = Path(r"C:\Users\latej\Desktop\Projets_github\nltk-lyrics-text-classification\data\stopword.txt")
STOP |= set(custom_path.read_text(encoding="utf-8").split())

print("Nb stopwords :", len(STOP))
```

## 3) Split train/test (stratifié)

On démarre avec un split stratifié classique (utile pour comparer rapidement).  
Ensuite, on fera un split **par artiste** pour tester la généralisation.


```python
from sklearn.model_selection import train_test_split

X = df["lyrics"]
y = df["genre"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

y_train.value_counts(normalize=True), y_test.value_counts(normalize=True)
```

## 4) Baseline (classe majoritaire)

Une baseline simple permet de contextualiser les scores des modèles.

```python

from sklearn.dummy import DummyClassifier
from sklearn.metrics import classification_report, accuracy_score

dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(X_train, y_train)
pred_dummy = dummy.predict(X_test)

print("Accuracy baseline:", accuracy_score(y_test, pred_dummy))
print(classification_report(y_test, pred_dummy, digits=4))
```

## 5) Comparaison de modèles (TF‑IDF mots)

On compare trois classiques très efficaces en classification de texte :
- Naive Bayes (baseline “texte”)
- Logistic Regression
- Linear SVM (LinearSVC)

On utilise un **Pipeline** scikit‑learn pour éviter toute fuite de données.


```python

import re
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import f1_score

def preprocessor(text: str) -> str:
    text = "" if text is None else str(text)
    text = text.lower()
    text = re.sub(r"http\S+|www\.\S+", " ", text)  # urls
    text = re.sub(r"\s+", " ", text).strip()
    return text

tfidf_word = TfidfVectorizer(
    preprocessor=preprocessor,
    stop_words=STOP,
    ngram_range=(1,2),
    min_df=2,
)

pipelines = {
    "MultinomialNB": Pipeline([("tfidf", tfidf_word), ("model", MultinomialNB())]),
    "LogReg": Pipeline([("tfidf", tfidf_word), ("model", LogisticRegression(max_iter=3000))]),
    "LinearSVC": Pipeline([("tfidf", tfidf_word), ("model", LinearSVC())]),
}

results = []
for name, pipe in pipelines.items():
    pipe.fit(X_train, y_train)
    pred = pipe.predict(X_test)
    results.append({
        "model": name,
        "accuracy": accuracy_score(y_test, pred),
        "f1_macro": f1_score(y_test, pred, average="macro"),
        "f1_rap": f1_score((y_test=="Rap"), (pred=="Rap")),
    })

import pandas as pd
res_df = pd.DataFrame(results).sort_values("f1_macro", ascending=False)
res_df
```

### Visualisation : comparaison des métriques

Un graphe permet de comparer rapidement les modèles.

```python

import matplotlib.pyplot as plt

plot_df = res_df.set_index("model")[["accuracy","f1_macro","f1_rap"]]
ax = plot_df.plot(kind="bar", rot=0)
ax.set_ylim(0,1)
ax.set_title("Comparaison des modèles (test)")
ax.set_ylabel("Score")
plt.tight_layout()
plt.show()
```

## 6) Évaluation détaillée du meilleur modèle

On prend le meilleur modèle (selon F1 macro) et on affiche :
- classification report
- matrice de confusion (visuelle)
- distribution des scores (pour comprendre la marge de décision)


```python

from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix

best_name = res_df.iloc[0]["model"]
best_model = pipelines[best_name]
best_model.fit(X_train, y_train)
pred_best = best_model.predict(X_test)

print("Meilleur modèle:", best_name)
print("Accuracy:", accuracy_score(y_test, pred_best))
print(classification_report(y_test, pred_best, digits=4))

cm = confusion_matrix(y_test, pred_best, labels=["VF","Rap"])
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["VF","Rap"])
disp.plot()
plt.title(f"Matrice de confusion — {best_name}")
plt.tight_layout()
plt.show()
```

### Visualisation : scores de décision (LinearSVC)

Pour LinearSVC, on peut analyser `decision_function` : ça montre si le modèle est “confiant” ou hésite.

```python

import matplotlib.pyplot as plt
import numpy as np

if best_name == "LinearSVC":
    scores = best_model.decision_function(X_test)  # positif => classe 1 selon l'ordre interne
    # retrouver la classe positive (ici scikit-learn encode classes_ par ordre alphabétique)
    classes = list(best_model.named_steps["model"].classes_)
    # on trace la distribution par vraie classe
    fig = plt.figure()
    plt.hist(scores[y_test.values == classes[0]], bins=40, alpha=0.6, label=f"Vrai: {classes[0]}")
    plt.hist(scores[y_test.values == classes[1]], bins=40, alpha=0.6, label=f"Vrai: {classes[1]}")
    plt.axvline(0, linestyle="--")
    plt.title("Distribution des scores de décision (LinearSVC)")
    plt.xlabel("decision_function")
    plt.ylabel("count")
    plt.legend()
    plt.tight_layout()
    plt.show()
else:
    print("Le graphe de scores est surtout pertinent pour LinearSVC.")
```

## 7) Interprétation : n-grams les plus discriminants

Sur un modèle linéaire (LogReg ou LinearSVC), on peut regarder les **coefficients** pour voir quels n‑grams poussent vers Rap ou VF.


```python

import numpy as np
import pandas as pd

model = best_model.named_steps["model"]
vectorizer = best_model.named_steps["tfidf"]

if hasattr(model, "coef_"):
    feature_names = np.array(vectorizer.get_feature_names_out())
    coefs = model.coef_.ravel()  # binaire
    top_pos = np.argsort(coefs)[-25:][::-1]
    top_neg = np.argsort(coefs)[:25]

    # classes_
    classes = list(model.classes_)
    # convention : pour binaire, coef_ correspond à classes_[1] vs classes_[0]
    pos_class = classes[1]
    neg_class = classes[0]

    top_df = pd.DataFrame({
        f"Top n-grams → {pos_class}": feature_names[top_pos],
        "coef+": coefs[top_pos],
        f"Top n-grams → {neg_class}": feature_names[top_neg],
        "coef-": coefs[top_neg],
    })
    print(top_df)
else:
    print("Ce modèle n'expose pas de coefficients interprétables (coef_).")
```

## 8) Analyse d’erreurs (exemples commentés)

On extrait quelques faux positifs / faux négatifs pour comprendre :
- quelles paroles VF ressemblent au Rap (et inversement),
- si la longueur, le vocabulaire ou le style expliquent les erreurs.

👉 Sur GitHub/CV, cette section est très différenciante.


```python

import pandas as pd

df_test = pd.DataFrame({"lyrics": X_test.values, "y_true": y_test.values, "y_pred": pred_best})
df_test["is_error"] = df_test["y_true"] != df_test["y_pred"]

fp = df_test[(df_test["y_true"]=="VF") & (df_test["y_pred"]=="Rap")].head(5)
fn = df_test[(df_test["y_true"]=="Rap") & (df_test["y_pred"]=="VF")].head(5)

def short(text, n=350):
    t = text.replace("\n"," ")
    return t[:n] + ("…" if len(t) > n else "")

print("Faux positifs (VF → Rap) :")
display(fp.assign(extrait=fp["lyrics"].apply(short))[["y_true","y_pred","extrait"]])

print("\nFaux négatifs (Rap → VF) :")
display(fn.assign(extrait=fn["lyrics"].apply(short))[["y_true","y_pred","extrait"]])
```

## 9) Validation plus réaliste : split par artiste (si disponible)

Dans un split aléatoire, on peut “mélanger” des chansons d’un même artiste entre train et test.  
Le split **par artiste** teste si le modèle généralise à des artistes jamais vus (plus réaliste).


```python

from sklearn.model_selection import GroupShuffleSplit

if "auteur" in df.columns:
    groups = df["auteur"].astype(str).values

    gss = GroupShuffleSplit(n_splits=1, test_size=0.30, random_state=42)
    train_idx, test_idx = next(gss.split(X, y, groups=groups))

    X_train_g, X_test_g = X.iloc[train_idx], X.iloc[test_idx]
    y_train_g, y_test_g = y.iloc[train_idx], y.iloc[test_idx]

    # ré-entraînement du meilleur modèle
    best_model.fit(X_train_g, y_train_g)
    pred_g = best_model.predict(X_test_g)

    print("Split par artiste —", best_name)
    print("Accuracy:", accuracy_score(y_test_g, pred_g))
    from sklearn.metrics import f1_score
    print("F1 macro:", f1_score(y_test_g, pred_g, average="macro"))
    print(classification_report(y_test_g, pred_g, digits=4))

    cm_g = confusion_matrix(y_test_g, pred_g, labels=["VF","Rap"])
    disp_g = ConfusionMatrixDisplay(confusion_matrix=cm_g, display_labels=["VF","Rap"])
    disp_g.plot()
    plt.title(f"Matrice de confusion (split artiste) — {best_name}")
    plt.tight_layout()
    plt.show()
else:
    print("Colonne 'artist' absente : impossible de faire un split par artiste.")
```

## 10) Optimisation légère (optionnelle)

On peut faire un tuning rapide (sans sur-optimiser) pour vérifier si on gagne un peu.  
On visualise ensuite l’impact sur les métriques.


```python

from sklearn.model_selection import RandomizedSearchCV

# on tune uniquement LinearSVC (souvent le meilleur) ; sinon on saute
if "LinearSVC" in pipelines:
    base_pipe = pipelines["LinearSVC"]

    param_distributions = {
        "tfidf__ngram_range": [(1,1),(1,2)],
        "tfidf__min_df": [1,2,3,5],
        "model__C": np.logspace(-2, 1.2, 30),  # ~0.01 -> ~15.8
    }

    search = RandomizedSearchCV(
        base_pipe,
        param_distributions=param_distributions,
        n_iter=25,
        scoring="f1_macro",
        cv=5,
        random_state=42,
        n_jobs=-1,
        verbose=0
    )

    search.fit(X_train, y_train)
    tuned = search.best_estimator_

    pred_tuned = tuned.predict(X_test)
    print("Meilleurs params:", search.best_params_)
    print("Test — Accuracy:", accuracy_score(y_test, pred_tuned))
    from sklearn.metrics import f1_score
    print("Test — F1 macro:", f1_score(y_test, pred_tuned, average="macro"))
    print(classification_report(y_test, pred_tuned, digits=4))

    # visuel : distribution des scores CV
    cvres = pd.DataFrame(search.cv_results_).sort_values("rank_test_score")
    plt.figure()
    plt.plot(cvres["mean_test_score"].values, marker="o")
    plt.title("RandomizedSearchCV — scores CV (F1 macro)")
    plt.xlabel("Essai (trié par rang)")
    plt.ylabel("mean CV score")
    plt.tight_layout()
    plt.show()

else:
    print("LinearSVC absent des pipelines.")
```

## 11) Courbe Precision–Recall (classe **Rap**)

Les données étant déséquilibrées (plus de VF que de Rap), la **courbe Precision–Recall** est souvent plus informative que la ROC.
Elle montre le compromis entre :
- **Precision** : parmi les chansons prédites Rap, combien sont réellement Rap ?
- **Recall** : parmi les chansons réellement Rap, combien sont détectées ?


```python
from sklearn.metrics import precision_recall_curve, average_precision_score

# On transforme y_test en binaire : Rap = 1, VF = 0
y_true_bin = (y_test.values == "Rap").astype(int)

# Scores continus nécessaires pour PR curve
if hasattr(best_model, "decision_function"):
    scores = best_model.decision_function(X_test)
elif hasattr(best_model, "predict_proba"):
    scores = best_model.predict_proba(X_test)[:, 1]
else:
    raise ValueError("Le modèle ne fournit ni decision_function ni predict_proba")

prec, rec, _ = precision_recall_curve(y_true_bin, scores)
ap = average_precision_score(y_true_bin, scores)

plt.figure()
plt.plot(rec, prec)
plt.title(f"Precision–Recall (classe Rap) — AP = {ap:.3f} — {best_name}")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.ylim(0, 1.05)
plt.xlim(0, 1.0)
plt.tight_layout()
plt.show()
```

## 12) Learning curve (le modèle a-t-il besoin de plus de données ?)

La **learning curve** permet de voir si la performance continue d’augmenter quand on donne plus de données au modèle.
- Si la courbe de validation **monte encore**, plus de données pourraient aider.
- Si elle **plafonne**, il faudra plutôt agir sur les features / le modèle / la qualité des données.


```python
from sklearn.model_selection import learning_curve

train_sizes, train_scores, val_scores = learning_curve(
    best_model,
    X,
    y,
    train_sizes=np.linspace(0.1, 1.0, 6),
    cv=5,
    scoring="f1_macro",
    n_jobs=-1,
    shuffle=True,
    random_state=42,
)

train_mean = train_scores.mean(axis=1)
train_std  = train_scores.std(axis=1)
val_mean   = val_scores.mean(axis=1)
val_std    = val_scores.std(axis=1)

plt.figure()
plt.plot(train_sizes, train_mean, marker="o", label="Train")
plt.fill_between(train_sizes, train_mean-train_std, train_mean+train_std, alpha=0.2)
plt.plot(train_sizes, val_mean, marker="o", label="Validation (CV)")
plt.fill_between(train_sizes, val_mean-val_std, val_mean+val_std, alpha=0.2)
plt.title(f"Learning curve — {best_name} (scoring = F1 macro)")
plt.xlabel("Nombre d'exemples d'entraînement")
plt.ylabel("F1 macro")
plt.legend()
plt.tight_layout()
plt.show()
```
